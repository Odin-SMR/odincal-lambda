```python
import numpy as np


def get_seq(
    mode: int,
):
    """get the ac chip configuration from the mode parameter"""
    seq = np.zeros(16, dtype=int)
    ssb = [1, -1, 1, -1, -1, 1, -1, 1]
    mode = (mode << 1) | 1
    for i in range(8):
        if mode & 1:
            m = i
        seq[2 * m] = seq[2 * m] + 1
        mode >>= 1

    for i in range(8):
        if seq[2 * i]:
            if ssb[i] < 0:
                seq[2 * i + 1] = -1
            else:
                seq[2 * i + 1] = 1
        else:
            seq[2 * i + 1] = 0
    return seq
```
