```python
import boto3

profile_name = "odin-cdk"
session = boto3.Session(profile_name=profile_name)
credentials = session.get_credentials()
```

```python
import json

import boto3

STATE_MACHINE = (
    "arn:aws:states:eu-north-1:991049544436:stateMachine:OdinSMROdincalStateMachine"
)


sfn_client = boto3.client(
    "stepfunctions",
    region_name="eu-north-1",
    aws_access_key_id=credentials.access_key,
    aws_secret_access_key=credentials.secret_key,
)
```

```python
from datetime import datetime, timezone

import awswrangler as wr


s3uri = "s3://odin-pdc-l0/ac*"
FILTER_DATE_START = datetime(2024, 9, 17, tzinfo=timezone.utc)
FILTER_DATE_STOP = datetime(2025, 1, 29, tzinfo=timezone.utc)

objects = wr.s3.list_objects(
    s3uri,
    boto3_session=session,
    last_modified_begin=FILTER_DATE_START,
    last_modified_end=FILTER_DATE_STOP,
)

for obj in objects:
    j = obj[17:]
    type = "ac2"
    if j.startswith("ac1"):
        type = "ac1"
    sfn_input = json.dumps(dict(name=j, type=type))
    print(sfn_input)
    sfn_client.start_execution(stateMachineArn=STATE_MACHINE, input=sfn_input)
```
